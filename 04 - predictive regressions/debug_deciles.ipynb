```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
# Directories
CODE_DIR = Path(r"C:\Users\willi\.vscode\Github\ml-from-crowd")
DATA_DIR = Path(r"E:\Research_data\Stocktwits\dataset\v1\data\csv")
FIGURES_DIR = Path(r"C:\Users\willi\.vscode\Github\Figures")
MODEL_DATA_DIR = Path(r"C:\Users\willi\.vscode\Github\Data")

# Load predictions for the 2-feature linear model
preds = pd.read_pickle(MODEL_DATA_DIR / "predictions_linear_regression_input=2.pkl")
preds['date'] = pd.to_datetime(preds['date'])

print(f"Shape: {preds.shape}")
print(f"Columns: {list(preds.columns)}")
print(f"Date range: {preds['date'].min().date()} to {preds['date'].max().date()}")
preds.head()
```

```python
# Pick a single date in January 2012
target_date = preds[preds['date'].dt.to_period('M') == '2012-01']['date'].unique()[0]
day = preds[preds['date'] == target_date].copy()

print(f"Date: {target_date.date()}")
print(f"Number of stocks: {len(day):,}")
print(f"\nPrediction distribution:")
print(day['prediction'].describe())
```

```python
# Number of unique prediction values
n_unique = day['prediction'].nunique()
print(f"Unique prediction values: {n_unique:,} out of {len(day):,} stocks")
print(f"\nValue counts of most common predictions:")
print(day['prediction'].value_counts().head(20))
```

```python
# Split into deciles using pd.qcut
day['decile'] = pd.qcut(day['prediction'].rank(method='first'), 10, labels=False) + 1

# Stocks per decile
print("Stocks per decile:")
print(day.groupby('decile').size().to_string())

# Breakpoints
print("\nPrediction range per decile:")
decile_stats = day.groupby('decile')['prediction'].agg(['min', 'max', 'mean', 'count'])
print(decile_stats.to_string())
```
